# **OmegaFold Installation and Usage Guide**

**1. Installation**

To install OmegaFold, follow the instructions available in the official GitHub repository:
👉 https://github.com/HeliXonProtein/OmegaFold

To prepare the environment and install OmegaFold directly from the source code:

```python
pip install git+https://github.com/HeliXonProtein/OmegaFold.git
```

Alternatively, you can clone the repository manually:

```python
git clone https://github.com/HeliXonProtein/OmegaFold
cd OmegaFold
python setup.py install
```

**2. Running OmegaFold**

To run OmegaFold, use the following command:

```python
omegafold INPUT_FILE.fasta OUTPUT_DIRECTORY
```

**NOTE:** The file INPUT_FILE.fasta should be a standard FASTA file, possibly containing multiple sequences, each preceded by a header line starting with > or : followed by the amino acid sequence.

**3. FASTA Input Files**

The FASTA sequences are available in different formats on my Google Drive (you can use any of them as input):

**4. Resource and Performance Notes**

You can adjust OmegaFold’s performance and memory usage by tuning the following parameters:

*   **Trade compute time for lower memory (GRAM) usage:**

    Use the parameter --subbatch_size.
    A smaller value reduces memory requirements but increases runtime.



*   **Trade compute time for prediction quality:**

    Adjust the parameter --num_cycle.
    Higher values generally improve model quality at the cost of longer runtime.

**5. Additional Information**

To view all available options and parameters:

```python
omegafold --help
```

**6. Alternative Execution (macOS Users)**

If installation fails, since OmegaFold relies on minimal third-party dependencies, you can manually install only PyTorch and Biopython.

For macOS users using the MPS accelerator, it may be necessary to install the latest nightly version of PyTorch.
In this case, you can run OmegaFold directly with:

```python
python main.py INPUT_FILE.fasta OUTPUT_DIRECTORY
```

**7. Output**

A separate PDB file is generated for each sequence in INPUT_FILE.fasta.
All resulting .pdb files are saved in the specified OUTPUT_DIRECTORY.
