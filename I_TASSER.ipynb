**6. Download Template Library Files**

Access the following page to download the necessary template libraries:
👉 https://zhanggroup.org/library/

**PDB**(~2099MB): non-redundant protein structures from PDB.

**MTX**(~737MB): Sequence profiles from PSI-BLAST search.

**DEP** (~441MB): Depth-dependent structure profiles of templates.

**SIG**(~466MB): Residue environmental profiles of templates.

**CNT** (~187MB): Side-chain center contacts of templates.

**dotProfile** (~243MB): Sequence profiles from PSI-BLAST search.

**stride**(~5MB): Seven state secondary structure assignment.

**summary1**(~82MB): Sequence, secondary structure, solvent accessibilty, phi/psi angles.

**summary2**(~185MB): Training data for phi/psi prediction and contact prediction by SVM.


**7. Extract the Files**

Extract each archive directly in the terminal:

```python
tar -xf archive.tar.bz2
```

If the extraction or download fails, try using the download script provided with I-TASSER:
